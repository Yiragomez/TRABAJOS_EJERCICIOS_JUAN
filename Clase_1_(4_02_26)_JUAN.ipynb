```python
class Automovil:
  def __init__(self, marca, modelo, color):#Constructor
    self.marca = marca #cuando colocamos (self) lo que hacemos es ir una por una
    self.modelo = modelo
    self.color = color

  def mostrar_info(self): #lee y evita colocar self una por una como en el ejemplo anterior es decir = METODO
    print(f"El marca del vehiculo: {self.marca}")
    print(f"El modelo del vehiculo: {self.modelo}")
    print(f"El color del vehiculo: {self.color}")

def main(): #cada que udt coloca main siempre va la definicion de todas las variables anteriores
    marca= input("Ingrese la marca del vehiculo: ") # Aqui podemos observar que estamos especificando una por una para que pidamos y al terminar nos bote el resultado esperado
    modelo= input("Ingrese el modelo del vehiculo: ")
    color= input("Ingrese el color del vehiculo: ")

    vehiculo1 = Automovil(marca, modelo, color)# Aqui ya lo envio para que se ejecute basandose al primer (def)
    vehiculo1.mostrar_info()

if __name__ == "__main__":
  main()




```

```python
class Perro:
  def __init__(self, tamano, raza, color, genero, edad ):#Constructor
    self.tamano = tamano
    self.raza = raza
    self.color = color
    self.genero = genero
    self.edad = edad
  def mostrar_info(self): #metodo
    print(f"El tamano del perro: {self.tamano}")
    print(f"La raza del perro: {self.raza}")
    print(f"El color del perro: {self.color}")
    print(f"El genero del perro: {self.genero}")
    print(f"La edad del perro: {self.edad}")
def main():
  tamano= input("Ingrese el tamano del perro: ")
  raza= input("Ingrese la raza del perro: ")
  color= input("Ingrese el color del perro: ")
  genero= input("Ingrese el genero del perro: ")
  edad= input("Ingrese la edad del perro: ")
  perro1 = Perro(tamano, raza, color, genero, edad)
  perro1.mostrar_info()

if __name__ == "__main__":
  main()

```

```python
class Avion:
  def __init__(self, tamano, color, modelo, ano, fabricante, materiales, motor):
    self.tamano = tamano
    self.color = color
    self.modelo = modelo
    self.ano = ano
    self.fabricante = fabricante
    self.materiales = materiales
    self.motor = motor
  def mostrar_info(self):
    print(f"El tamano del avion: {self.tamano}")
    print(f"El color del avion: {self.color}")
    print(f"El modelo del avion: {self.modelo}")
    print(f"El ano del avion: {self.ano}")
    print(f"El fabricante del avion: {self.fabricante}")
    print(f"Los materiales del avion: {self.materiales}")
    print(f"El motor del avion: {self.motor}")
def main():
  tamano= input("Ingrese el tamano del avion: ")
  color= input("Ingrese el color del avion: ")
  modelo= input("Ingrese el modelo del avion: ")
  ano= input("Ingrese el ano: ")
  fabricante= input("Ingrese el fabricante: ")
  materiales= input("Ingrese los materiales: ")
  motor= input("Ingrese el motor: ")
  avion1 = Avion(tamano, color, modelo, ano, fabricante, materiales, motor)
  avion1.mostrar_info()
if __name__ == "__main__":
  main()



```
