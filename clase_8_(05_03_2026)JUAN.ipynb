```python
class Operaciones():
    def __init__(self,num1,num2):
      self.num1=num1
      self.num2=num2
    def calcular(self):
      print("*****ALERTA VAMOS A INICIAR OPERACION ¡ATENTOS!*****")
class Suma(Operaciones):
    def __init__(self,num1,num2):
      super().__init__(num1,num2)#aqui los utilizo los atributos del padre
      print("Preparando la operacion suma")
    def calcular(self):
      super().calcular()
      resultado = self.num1 + self.num2
      print(f"El resultado de la suma es: {resultado}")
class Resta(Operaciones):
    def __init__(self,num1,num2):
      super().__init__(num1,num2)
      print("Preparando la operacion resta")
    def calcular(self):
      super().calcular()
      resultado1 = self.num1 - self.num2
      print(f"El resultado de la resta es: {resultado1}")
class Multiplicacion(Operaciones):
    def __init__(self,num1,num2):
      super().__init__(num1,num2)
      print("Preparando la operacion multiplicacion")
    def calcular(self):
      super().calcular()
      resultado = self.num1 * self.num2
      print(f"El resultado de la multiplicacion es: {resultado}")
class Division(Operaciones):
    def __init__(self,num1,num2):
      super().__init__(num1,num2)
    def calcular(self):
      super().calcular()
      resultado = self.num1 / self.num2
      print(f"El resultado de la division es: {resultado}")
class Potencia(Operaciones):
    def __init__(self,num1,num2):
      super().__init__(num1,num2)
      print("Preparando la operacion potencia")
    def calcular(self):
      super().calcular()
      resultado = self.num1 ** self.num2
      print(f"El resultado de la potencia es: {resultado}")



def main():
  operacion1 = Suma(5,3)
  operacion1.calcular()
  print("Operacion finalizada")

  operacion2 = Resta(10,3)
  operacion2.calcular()
  print("Operacion finalizada")

  operacion3 = Multiplicacion(8,6)
  operacion3.calcular()
  print("Operacion finalizada")

  operacion4 = Division(35,3)
  operacion4.calcular()
  print("Operacion finalizada")

  operacion5 = Potencia(45,3)
  operacion5.calcular()
if __name__ == "__main__":
  main()
  print("Fin del programa")


```
