```python
class Libro:
  def __init__(self, titulo, precio):
    self.__titulo = titulo #esta encapsulado
    self.__precio = precio

  @property #DECORADORES
  def titulo(self):
    return self.__titulo

  @property #DECORADORES
  def precio(self):
    return self.__precio


  @titulo.setter #DECORADORES
  def titulo(self, nuevo_titulo):
    if isinstance(nuevo_titulo, str):
      self.__titulo = nuevo_titulo
    else:
      print("Debe ingresar un titulo valido")

  @precio.setter #DECORADORES
  def precio(self, nuevo_precio):
    if isinstance(nuevo_precio, (int, float)) and nuevo_precio > 0:
      self.__precio = nuevo_precio
    else:
      print("Debe ingresar un numero valido en el precio")

#Aqui ya estamos mostrando la informacion a base delo que pedimos
  def mostrar_informacion(self):
    print(f"el Titulo del libro: {self.__titulo}")
    print(f"el Precio del libro: {self.__precio}")#o
    print(f"\nel Titulo del libro: {self.titulo}, El precio del Libro: {self.precio}")#Es igual pero mas resumido

def main():
  print("=====CREACION DE OBJETOS EN POO=====")
  libro1 = Libro("EL PRINCIPITO", 120000) #CREACION DE MI PRIMER OBJETO
  libro2 = Libro("LA DIVINA COMEDIA", 100000) #CREACION DE MI SEGUNDO OBJETO

  print("el nombre del primer libro es: ")
  print(libro1.titulo)
  print("el precio del  libro es: ",libro1.precio)
  print("El nombre del libro es: ",libro1.titulo, "Precio: ", libro1.precio)

  print("todos los datos")
  libro1.mostrar_informacion()

  print("\nActualizaremos el titulo")
  libro1.titulo =input("Ingrese el nuevo titulo ==")
  print("Titulo actualizado: ahora el libro 1 se llama: ", libro1.titulo)

  print("el precio del libro es: ",libro1.precio," Actualizaremos este precio por otro ")
  libro1.precio = input("Ingrese el nuevo precio ==")
  print("Precio actualizado: ahora el libro 1 cuesta: ", libro1.precio)
  print("todos los datos")
  libro1.mostrar_informacion()

if __name__ == "__main__": #para poder mostrar los valores en el escritorio
    main()





```
